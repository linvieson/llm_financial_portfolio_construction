```python
def create_dfs(read_path, write_path, n_remove_rows):
    df = pd.read_csv(read_path)
    df = df.drop(df.index[:n_remove_rows])
    df = df[['Ticker', 'Name', 'Sector', 'Industry']]
    df = df.rename({'Ticker': 'date', 'Name': 'ticker', 'Sector': 'close', 'Industry': 'marketcap'}, axis=1)

    reshaped_df = df.pivot(index='date', columns='ticker', values='close')
    reshaped_df = reshaped_df.reset_index()

    reshaped_df.to_csv(write_path)
```
