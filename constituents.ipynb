```python
wikiurl="https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
response = requests.get(wikiurl)
print(response.status_code)
```

```python
# parse data from the html into a beautifulsoup object
soup = BeautifulSoup(response.text, 'html.parser')
indiatable = soup.find_all('table',{'class':"wikitable"})
```

```python
df = pd.read_html(str(indiatable[0]))
df = pd.DataFrame(df[0])
df.head()
```

```python
sectors = ['Information Technology', 'Health Care', 'Financials', 'Consumer Discretionary', \
           'Communication Services', 'Industrials', 'Consumer Staples', 'Energy',\
           'Utilities', 'Real Estate', 'Materials']
```

```python
for sector in sectors:
    num = len(df[df['GICS Sector'] == sector])
    print(f'{sector} num of constituents: {num}')
```
